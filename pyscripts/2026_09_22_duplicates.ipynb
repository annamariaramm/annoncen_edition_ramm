```python
from pathlib import Path
import xml.etree.ElementTree as ET
from collections import defaultdict
import csv

# --------------------------------------------------
# 1. Ordner mit deinen TEI-Dateien
# --------------------------------------------------

tei_folder = Path("tei")

# --------------------------------------------------
# 2. TEI-Namespace
# --------------------------------------------------

NS = {
    "tei": "http://www.tei-c.org/ns/1.0"
}

# --------------------------------------------------
# 3. Anzeigen sammeln
# --------------------------------------------------

anzeigen = defaultdict(list)

for xml_file in tei_folder.glob("*.xml"):

    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()

    except ET.ParseError as e:
        print(f"Fehler beim Lesen von {xml_file}: {e}")
        continue

    # <text>/<body>/<div>/<p> suchen
    p = root.find(".//tei:text/tei:body/tei:div/tei:p", NS)

    if p is None:
        print(f"Kein <p> gefunden in: {xml_file.name}")
        continue

    # Gesamten Text aus <p> und allen darin enthaltenen
    # Elementen holen
    text = "".join(p.itertext())

    # Leerzeichen und Zeilenumbrüche vereinheitlichen
    text = " ".join(text.split())

    # Leere Texte überspringen
    if not text:
        print(f"Leerer Text in: {xml_file.name}")
        continue

    # Zum Testen ausgeben
    print(f"\n{xml_file.name}")
    print(text)

    # Text und Datei speichern
    anzeigen[text].append(xml_file.name)


# --------------------------------------------------
# 4. CSV mit Duplikaten erstellen
# --------------------------------------------------

with open(
    "identische_anzeigen.csv",
    "w",
    newline="",
    encoding="utf-8-sig"
) as csvfile:

    writer = csv.writer(csvfile)

    writer.writerow([
        "duplicate_group",
        "advertisement"
    ])

    group = 1

    for text, files in anzeigen.items():

        if len(files) > 1:

            for file in files:
                writer.writerow([
                    group,
                    file
                ])

            group += 1

print("Die Ergebnisse wurden gespeichert als: identische_anzeigen.csv")
```
